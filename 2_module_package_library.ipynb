```python
import math
import random
import datetime
import os
import sys
import time

print("### 파이썬 내장 모듈 예시 실행 ###\n")

# 1. math: 수학 함수
print("1. math 모듈")
math_result = math.sqrt(16)
print(f"math.sqrt(16) 실행 결과: {math_result}\n")

# 2. random: 랜덤값 생성
print("2. random 모듈")
random_result = random.randint(1, 10)
print(f"random.randint(1, 10) 실행 결과: {random_result} (1~10 사이의 무작위 정수)\n")

# 3. datetime: 날짜와 시간 처리
print("3. datetime 모듈")
datetime_result = datetime.datetime.now()
print(f"datetime.datetime.now() 실행 결과: {datetime_result}\n")

# 4. os: 운영체제 관련 기능
print("4. os 모듈")
os_result = os.getcwd()
print(f"os.getcwd() 실행 결과: {os_result} (현재 작업 중인 디렉토리 경로)\n")

# 5. sys: 시스템 관련 정보
print("5. sys 모듈")
sys_result = sys.version
print(f"sys.version 실행 결과:\n{sys_result}\n")

# 6. time: 시간 관련 기능
print("6. time 모듈")
print("time.sleep(1) 실행... (1초간 대기합니다)")
time.sleep(1)
print("1초 대기 완료!")
```

```python
import pandas as pd

print("=== 1. DataFrame 생성 ===")
# 사전(dict) 데이터를 기반으로 데이터프레임 생성
data = {
    "이름": ["김철수", "이영희", "박민수", "정지원"],
    "나이": [25, 30, 22, 28],
    "부서": ["개발", "디자인", "개발", "마케팅"],
    "점수": [88, 92, 75, 95]
}
df = pd.DataFrame(data)
print(df)

print("\n=== 2. 기본 정보 및 구조 확인 ===")
print("• 데이터 상단 2행:")
print(df.head(2))

print("\n• 데이터프레임 정보:")
df.info()

print("\n• 수치형 데이터 요약 통계:")
print(df.describe())

print("\n=== 3. 데이터 선택 및 필터링 ===")
# 특정 열 선택
print("• '이름' 열만 선택:")
print(df["이름"])

# 조건 검색 (점수가 85점 이상인 행)
print("\n• 점수가 85점 이상인 사람:")
high_scores = df[df["점수"] >= 85]
print(high_scores)

# loc 사용 (인덱스 및 열 이름 기반 접근)
print("\n• 첫 번째 행의 이름과 점수:")
print(df.loc[0, ["이름", "점수"]])

print("\n=== 4. 데이터 추가 및 수정 ===")
# 새로운 열 추가 (합격 여부)
df["합격여부"] = df["점수"].apply(lambda x: "합격" if x >= 80 else "불합격")

# 열 데이터 수정 (모든 사람의 나이 +1)
df["나이"] = df["나이"] + 1

print(df)

print("\n=== 5. 그룹화 및 통계 (groupby) ===")
# 부서별 평균 점수 구하기
dept_mean = df.groupby("부서")["점수"].mean()
print("• 부서별 평균 점수:")
print(dept_mean)
```
