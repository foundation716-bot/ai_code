```python
#정의
class student:
    def __init__(self, name, age):
        self.name = name+'ㅁ'
        self.age = age

    def introduce(self):
        print(f'안녕하세요 제 이름은 {self.name}입니다. {self.age}살입니다.')
        
```

```python
#클래스 정의
class Calculator:
    def __init__(self, number_list):
        self.number_list = number_list

    def sum(self):
        return sum(self.number_list)

    def avg(self):
        total = self.sum()
        return total / len(self.number_list)
```

```python
#객체 생성
cal1 = Calculator([1,2,3,4,5]) 

cal1.sum() # 15 

cal1.avg() # 3.0



cal2 = Calculator([6,7,8,9,10]) 

cal2.sum() # 40 

cal2.avg() #8.0

print(cal1.sum())
print(cal1.avg())
print(cal2.sum())
print(cal2.avg())
print(cal1.number_list)
print(cal2.number_list)
print(cal1.__dict__)
```

```python
import random

class Account:
    # 클래스 변수: 모든 계좌의 은행 이름은 SC은행으로 고정
    bank_name = "SC은행"

    def __init__(self, name, balance):
        self.name = name          # 예금주
        self.balance = balance    # 초기 잔액
        
        # 계좌번호 자동 생성 (3자리-2자리-6자리)
        num1 = str(random.randint(0, 999)).zfill(3)
        num2 = str(random.randint(0, 99)).zfill(2)
        num3 = str(random.randint(0, 999999)).zfill(6)
        
        self.account_number = f"{num1}-{num2}-{num3}"

    def display_info(self):
        print(f"은행이름: {self.bank_name}")
        print(f"예금주: {self.name}")
        print(f"계좌번호: {self.account_number}")
        print(f"잔액: {self.balance:,}원")
        print("-" * 30)

    

# --- 샘플 데이터 생성 및 테스트 ---

# 1. 예금주와 초기 잔액만 입력하여 계좌 객체 생성
acc1 = Account("홍길동", 100000)
acc2 = Account("김철수", 500000)
acc3 = Account("이영희", 2500000)

# 2. 계좌 정보 출력
acc1.display_info()
acc2.display_info()
acc3.display_info()

```

```python
# 1. 커스텀 예외 클래스 정의
class AgeException(Exception):
    def __init__(self, *args):
        # 부모 클래스(Exception)의 __init__에 *args를 전달하여 te.args 속성 생성
        super().__init__(*args)

# 2. 나이 입력 및 검증 함수
def input_age():
    age = int(input("나이를 입력하세요: "))
    
    # 입력값 체크
    if age <= 0:  # 조건 1: 0 이하의 음수나 0 입력 시
        raise AgeException("나이는 양수여야 합니다.")
        
    elif age > 150:  # 조건 2: 상한선(예: 150세) 초과 시
        raise AgeException("유효한 나이 범위를 초과했습니다. (150세 이하)")
        
    else:
        return age

# 3. 함수 실행 및 예외 처리
try:
    user_age = input_age()
    print(f"입력된 나이: {user_age}세")

except AgeException as te:
    # 예외 발생 시 전달한 첫 번째 메시지 출력 (te.args[0])
    print(f"[입력 오류] {te.args[0]}")
    
except ValueError:
    # 숫자가 아닌 문자열을 입력했을 때 예외 처리
    print("[입력 오류] 숫자만 입력해주세요.")
```
