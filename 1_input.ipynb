```python
snack_box = ['초코파이', '사탕', '껌', '초코바']

count = int(input("몇 상자를 주문하시겠어요? (숫자로): "))

print(f"\n총 {count}상자를 주문하셨습니다.")
for i in range(1, count + 1):
    print(f"{i}번째 상자 내용물: {snack_box}")
```

```python
#빼고 싶은 간식이 있으면 번호를 입력하세요

snack_box = ['초코파이', '사탕', '껌', '초코바']

# 1. 현재 간식 목록과 번호 출력
print("=== 현재 간식 목록 ===")
for i, snack in enumerate(snack_box):
    print(f"{i + 1}. {snack}")

# 2. 사용자로부터 빼고 싶은 간식 번호 입력받기
num = int(input("\n빼고 싶은 간식이 있으면 번호를 입력하세요: "))

# 3. 입력한 번호의 간식 제거 (인덱스는 0부터 시작하므로 입력값에서 1을 뺌)
if 1 <= num <= len(snack_box):
    removed = snack_box.pop(num - 1)
    print(f"\n'{removed}'이(가) 삭제되었습니다.")
else:
    print("\n올바른 번호를 입력하지 않았습니다.")

# 4. 최종 간식 목록 출력
print(f"남은 간식 상자: {snack_box}")

#코드 설명
enumerate(): 리스트의 요소와 함께 1번부터 시작하는 순번을 화면에 보여주기 위해 사용했습니다.
num - 1: 사용자는 1번부터 선택하지만, 파이썬 리스트의 인덱스는 0부터 시작하므로 입력받은 번호에서 1을 빼줍니다.
pop(): 해당 인덱스 위치의 요소를 리스트에서 삭제함과 동시에 어떤 항목이 삭제되었는지 알려줍니다.
#
```

```python
sentence = input("문장을 입력하세요: ")
words = ["파이썬", "프로그래밍", "언어"]
found_word = None
for word in words:
 if word in sentence:
  found_word = word
  break

if found_word:
 print(f"문장에 '{found_word}'가 포함되어 있습니다.")    
else:
 print("문장에 해당 단어가 포함되어 있지 않습니다.")
 
```
